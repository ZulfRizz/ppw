```python
import numpy as np
import pandas as pd
from scipy.sparse import load_npz

# 1. Baca matriks TF-IDF
X = load_npz("./ppw/tfidf_sparse.npz")

# 2. Baca nama fitur
with open("./ppw/tfidf_features.txt", "r", encoding="utf-8") as f:
    feature_names = [line.strip() for line in f]

# 3. Baca informasi dokumen dan label
docs = pd.read_csv("./ppw/tfidf_docs.csv")

print("Ukuran TF-IDF :", X.shape)
print("Jumlah fitur  :", len(feature_names))
print("Ukuran docs   :", docs.shape)
```

```python
df_tfidf = pd.DataFrame(
    X.toarray(),
    columns=feature_names
)

# Tambahkan informasi dokumen
df_tfidf.insert(0, "id", docs["id"].values)

# Tambahkan label
df_tfidf["label"] = docs["label"].values

print(df_tfidf.shape)
print(df_tfidf.head())
```

```python
df_tfidf.to_csv(
    "hasil_tfidf.csv",
    index=False
)
```
